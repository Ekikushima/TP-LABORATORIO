3.1. Carga y exploración inicial:

Utilizamos la librería pandas para leer los archivos .csv. Mediante un bucle iterativo, extraemos para cada tabla sus dimensiones, tipos de datos, cantidad de valores nulos y duplicados, estadísticas descriptivas básicas, y una muestra de las primeras y últimas filas para confirmar la correcta lectura de los registros.


```python
import pandas as pd
from IPython.display import display

# Definir los archivos del dataset
archivos = {
    'Actividad (Hechos)': 'data/Fact_User_Activity.csv',
    'Usuarios (Dimensión)': 'data/Dim_Users.csv',
    'Estado Mental (Dimensión)': 'data/Dim_MentalState.csv',
    'Hábitos (Dimensión)': 'data/Dim_Habits.csv',
    'Fechas (Dimensión)': 'data/Dim_Date.csv'
}

# Diccionario para almacenar los DataFrames en memoria
dfs = {}

# Bucle para cumplir con 3.1 a, b y c
for nombre, archivo in archivos.items():
    # 3.1.a - Cargar el dataset
    df = pd.read_csv(archivo)
    dfs[nombre] = df

    print(f"\n{'='*60}")
    print(f" TABLA: {nombre} ".center(60, '='))
    print(f"{'='*60}")

    # 3.1.b - Describir la estructura
    print(f"\n[Estructura] Dimensiones (filas, columnas): {df.shape}")
    print("\n[Tipos de dato por columna]:")
    print(df.dtypes)

    # 3.1.c - Analizar calidad (nulos y duplicados)
    print("\n[Calidad de Datos]")
    print(f"Valores nulos totales: {df.isnull().sum().sum()}")
    print(f"Filas duplicadas: {df.duplicated().sum()}")

    # 3.1.c - Estadísticas descriptivas (solo para columnas numéricas)
    print("\n[Estadísticas Descriptivas]:")
    display(df.describe())

    # 3.1.a - Primeras y últimas filas
    print("\n[Muestra de datos - Primeras 3 filas]:")
    display(df.head(3))
    print("\n[Muestra de datos - Últimas 3 filas]:")
    display(df.tail(3))
```

Comentarios y Hallazgos de la Exploración Inicial:

Calidad de los datos: No se detectaron filas duplicadas en ninguna de las 5 tablas. Se encontró un único valor nulo en la tabla Dim_Habits (en la columna Coffee_Level), el cual no afecta nuestro análisis principal.

Estructura y Tipos: Los tipos de datos (int64, float64, object) han sido inferidos correctamente por Pandas. Las claves de conexión (Keys) están en formato entero, lo que facilitará los cruces (merges) posteriores.

Cobertura Temporal: La tabla Dim_Date abarca los 365 días del año 2026.

Coherencia Descriptiva: En Fact_User_Activity, el promedio de Study_Hours y Total_Reels_Watched muestran valores lógicos para una población estudiantil, permitiendo avanzar con confianza hacia la formulación de la pregunta de investigación.

**3.2. Pregunta de investigación**


a)"¿En qué medida la intensidad de consumo de videos cortos se relaciona con una disminución en la capacidad de retención y un mayor deterioro cognitivo, afectando consecuentemente el rendimiento académico?"

b) ¿Por qué vale la pena responderla?

Demostrar esta relación permite visibilizar el problema del "deterioro cognitivo" (brain rot) y tomar decisiones informadas sobre el uso de la tecnología y la salud mental en el ámbito educativo.



3.3. Transformaciones y resumen estadístico
Transformaciones realizadas:

Merge (Unión de tablas): Aplicamos un pd.merge() para unir la tabla de hechos de actividad con la dimensión de estado mental mediante la clave StateKey. Esto es indispensable para cruzar los videos vistos con el nivel de deterioro cognitivo.

Binning (Discretización): Utilizamos pd.qcut() sobre la variable Total_Reels_Watched para crear una nueva columna categórica (Nivel_Consumo_Videos). Esto nos permite comparar estadísticamente a los usuarios agrupándolos en consumo Bajo, Medio y Alto.

```python
# 1. Primera transformación: Merge (Unión)
df_completo = pd.merge(dfs['Actividad (Hechos)'], dfs['Estado Mental (Dimensión)'], on='StateKey', how='inner')

# 2. Segunda transformación: Binning (Categorización)
df_completo['Nivel_Consumo_Videos'] = pd.qcut(df_completo['Total_Reels_Watched'], q=3, labels=['Bajo', 'Medio', 'Alto'])

# 3. Resumen estadístico agrupado (Groupby)
# Calculamos el promedio (mean) de horas de estudio y sesiones de concentración
resumen_estadistico = df_completo.groupby('Nivel_Consumo_Videos', observed=False)[['Study_Hours', 'Focus_Sessions_Count']].mean()

# Mostrar el resultado final
print("--- Impacto del consumo de videos cortos en el rendimiento académico ---")
display(resumen_estadistico)
```

**3.4. Análisis gráfico**
Para visualizar el impacto del consumo de videos cortos en el rendimiento académico, generaremos tres visualizaciones clave utilizando matplotlib y seaborn:

Gráfico de Cajas (Boxplot): Para analizar la distribución de las horas de estudio según la etapa de deterioro cognitivo (Brainrot Stage).

Gráfico de Barras (Barplot): Para comparar el promedio de sesiones de concentración logradas según el nivel de consumo de videos.

Diagrama de Dispersión (Scatterplot): Para observar la correlación directa entre el volumen total de videos vistos y las horas de estudio diarias.



```python
import matplotlib.pyplot as plt
import seaborn as sns

# Configurar el estilo visual
sns.set_theme(style="whitegrid")

# ---------------------------------------------------------
# Gráfico 1: Distribución (Boxplot)
# ---------------------------------------------------------
plt.figure(figsize=(10, 6))

sns.boxplot(
    data=df_completo,
    x='Brainrot_Stage',
    y='Study_Hours',
    hue='Brainrot_Stage',
    palette='Set2',
    legend=False
)

plt.title('Distribución de Horas de Estudio según Nivel de Deterioro Cognitivo', fontsize=14)
plt.xlabel('Etapa de Deterioro (Brainrot Stage)', fontsize=12)
plt.ylabel('Horas de Estudio Diarias', fontsize=12)

# Guardar como imagen
plt.savefig('grafico1_distribucion.png', bbox_inches='tight')
plt.show()
```

Gráfico 1: El diagrama de caja revela que los usuarios en etapas avanzadas de deterioro cognitivo (Critical / Advanced) presentan medianas de horas de estudio significativamente más bajas y menor dispersión hacia arriba, indicando un "techo" limitante en su capacidad académica.

```python
# ---------------------------------------------------------
# Gráfico 2: Comparación (Gráfico de barras)
# ---------------------------------------------------------
plt.figure(figsize=(8, 5))

sns.barplot(
    data=df_completo,
    x='Nivel_Consumo_Videos',
    y='Focus_Sessions_Count',
    hue='Nivel_Consumo_Videos',
    palette='Set2',
    legend=False
)

plt.title('Promedio de Sesiones de Concentración por Nivel de Consumo de Videos', fontsize=12)
plt.xlabel('Nivel de Consumo de Videos Cortos', fontsize=12)
plt.ylabel('Cantidad de Sesiones de Concentración', fontsize=12)

# Guardar como imagen
plt.savefig('grafico2_barras.png', bbox_inches='tight')
plt.show()
```

Gráfico 2: El gráfico de barras ratifica la tabla resumen anterior: existe una caída drástica y escalonada en la capacidad de mantener el enfoque. Pasar de un consumo "Bajo" a uno "Alto" reduce las sesiones de concentración casi a cero.

```python
# ---------------------------------------------------------
# Gráfico 3: Relación (Diagrama de dispersión)
# ---------------------------------------------------------
plt.figure(figsize=(10, 6))
sns.scatterplot(data=df_completo, x='Total_Reels_Watched', y='Study_Hours', alpha=0.5, color='coral')
plt.title('Relación entre Videos Vistos y Horas de Estudio', fontsize=14)
plt.xlabel('Cantidad Total de Reels/Shorts Vistos', fontsize=12)
plt.ylabel('Horas de Estudio Diarias', fontsize=12)
# Guardar como imagen
plt.savefig('grafico3_dispersion.png', bbox_inches='tight')
plt.show()
```

Gráfico 3: La dispersión muestra una clara tendencia negativa. A medida que aumenta el volumen de scroll en el eje X, las horas de estudio posibles en el eje Y descienden rápidamente, concentrándose en la parte inferior derecha del cuadrante.

La evidencia empírica del dataset confirma que una alta intensidad en el consumo de videos cortos está fuertemente relacionada con el deterioro cognitivo (Brainrot). Este fenómeno impacta negativamente en el rendimiento académico, reduciendo las horas de estudio a la mitad (de 5.6 a 2.4 horas) y destruyendo la capacidad de retención y enfoque de los estudiantes afectados.
